### Classification

```python
# Basic libraries
import numpy as np
import pandas as pd

# Scikit-learn modules
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import load_breast_cancer
```

```python
# Load dataset
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target
```

```python
X
```

```python
y
```

```python
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
# Base estimator: Decision Tree (stump)
base_estimator = DecisionTreeClassifier(max_depth=1)
```

```python
# Create AdaBoost classifier
adaboost = AdaBoostClassifier(estimator=base_estimator, n_estimators=50, learning_rate=1.0, random_state=42)
```

```python
adb = adaboost.fit(X_train,y_train)
```

```python
adb.score(X_test, y_test)
```

```python
y_pred = adb.predict(X_test)
```

```python
y_pred
```

```python
# Evaluate
print("Accuracy (Basic Model):", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
# Parameter grid
param_grid = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 1, 2]
}
```

```python
# AdaBoost with Decision Tree as base estimator
ada = AdaBoostClassifier(estimator=DecisionTreeClassifier(), random_state=42)
```

```python
# GridSearchCV
grid_search = GridSearchCV(estimator=ada,
                           param_grid=param_grid,
                           cv=5,            # 5-fold cross-validation
                           scoring='accuracy',
                           n_jobs=-1,        # Use all CPU cores
                           verbose=2)
```

```python
grid_search.fit(X_train,y_train)
```

```python
# Best parameters
print("Best Parameters:", grid_search.best_params_)
```

```python
print("Best Score on Training Set:", grid_search.best_score_)
```

```python
# Best estimator from GridSearch
best_ada = grid_search.best_estimator_

# Predict
y_pred_best = best_ada.predict(X_test)

# Evaluate
print("Accuracy (Tuned Model):", accuracy_score(y_test, y_pred_best))
print("Classification Report:\n", classification_report(y_test, y_pred_best))
```

### Regression

```python
# ══════════════════════════════════════════════════════════
# STEP 0 : Import all libraries
# ══════════════════════════════════════════════════════════

import numpy as np                        # for numerical operations like sqrt, array
import pandas as pd                       # for creating and handling DataFrames
import matplotlib                         # base matplotlib — needed to set backend
matplotlib.use("Agg")                     # use non-interactive backend (saves to file, no popup window)
import matplotlib.pyplot as plt           # for plotting graphs
import seaborn as sns                     # for beautiful statistical plots (barplot etc.)
import joblib                             # for saving and loading sklearn models efficiently
import warnings                           # to control warning messages
warnings.filterwarnings("ignore")         # suppress all warnings to keep output clean
```

```python
from sklearn.datasets import load_diabetes          # built-in diabetes dataset (no internet needed)
from sklearn.model_selection import train_test_split   # splits data into train and test sets
from sklearn.model_selection import RandomizedSearchCV # random hyperparameter search with cross-validation
from sklearn.model_selection import cross_val_score    # compute CV score on any model
from sklearn.ensemble import AdaBoostRegressor         # the main AdaBoost model for regression
from sklearn.tree import DecisionTreeRegressor         # weak learner used inside AdaBoost
from sklearn.preprocessing import StandardScaler       # scales features to mean=0, std=1
from sklearn.metrics import mean_absolute_error        # MAE metric — avg absolute difference
from sklearn.metrics import mean_squared_error         # MSE metric — used to compute RMSE
from sklearn.metrics import r2_score                   # R2 metric — how well model explains variance
from scipy.stats import randint                        # random integer distribution for param search
from scipy.stats import uniform                        # random float distribution for param search
```

```python
# ══════════════════════════════════════════════════════════
# STEP 1 : Load Dataset
# Diabetes dataset : 442 patients x 10 features
# Target : disease progression score (continuous, 25–346)
# ══════════════════════════════════════════════════════════

diabetes = load_diabetes()                                      # load the full diabetes dataset object from sklearn

df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)  # convert feature array to DataFrame with column names
df["DiseaseProgress"] = diabetes.target                            # add target column (disease score) to same DataFrame

print("Step 1 — Dataset loaded")
print(f"   Shape    : {df.shape}")                         # print rows x columns
print(f"   Features : {list(diabetes.feature_names)}")     # print list of feature names
print(f"   Target   : DiseaseProgress (continuous disease score)")
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 2 : Explore the Data
# ══════════════════════════════════════════════════════════

print("Step 2 — Data exploration")
print(df.describe().round(3).T.to_string())   # show count/mean/std/min/max for every column, transposed for readability
print()
print("   Missing values:")
print(df.isnull().sum().to_string())          # count NaN values per column — 0 means no missing data
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 3 : Outlier Check on Target (IQR method)
# IQR = Inter Quartile Range = Q3 - Q1
# Any value outside [Q1 - 1.5*IQR, Q3 + 1.5*IQR] = outlier
# ══════════════════════════════════════════════════════════

Q1 = df["DiseaseProgress"].quantile(0.25)    # 25th percentile — bottom of middle 50%
Q3 = df["DiseaseProgress"].quantile(0.75)    # 75th percentile — top of middle 50%
IQR = Q3 - Q1                                # inter-quartile range = spread of middle 50% of data

lower_bound = Q1 - 1.5 * IQR                # anything below this is a suspected outlier
upper_bound = Q3 + 1.5 * IQR                # anything above this is a suspected outlier

# count rows where target is outside the acceptable range (both conditions checked with OR)
outlier_count = ((df["DiseaseProgress"] < lower_bound) | (df["DiseaseProgress"] > upper_bound)).sum()

print("Step 3 — Outlier check on target")
print(f"   Q1={Q1:.1f}  Q3={Q3:.1f}  IQR={IQR:.1f}")
print(f"   Bounds : [{lower_bound:.1f}, {upper_bound:.1f}]")     # safe range for target values
print(f"   Outliers found : {outlier_count} (kept — AdaBoost is robust)")
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 4 : Separate Features (X) and Target (y)
# X = input columns (what we give the model)
# y = output column (what we want the model to predict)
# ══════════════════════════════════════════════════════════

X = df.drop(columns=["DiseaseProgress"])   # X = all columns EXCEPT target → input features
y = df["DiseaseProgress"]                  # y = only the target column → what we predict

print(f"Step 4 — Features and target separated")
print(f"   X shape : {X.shape}")                                                         # should be (442, 10)
print(f"   y shape : {y.shape},  min={y.min():.0f}, max={y.max():.0f}, mean={y.mean():.1f}")  # range check on target
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 5 : Train/Test Split  80% train, 20% test
# train set = model learns from this
# test set  = model never sees this during training
# ══════════════════════════════════════════════════════════

X_train, X_test, y_train, y_test = train_test_split(
    X, y,            # pass features and target
    test_size=0.2,   # 20% of data goes to test set, 80% to train
    random_state=42  # fixed seed so split is same every run (reproducibility)
)

print("Step 5 — Train/Test split done")
print(f"   Train rows : {X_train.shape[0]}")   # how many samples for training
print(f"   Test rows  : {X_test.shape[0]}")    # how many samples for testing
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 6 : Feature Scaling (StandardScaler)
# Transforms each feature to mean=0, std=1
# IMPORTANT: fit ONLY on train data to avoid data leakage
# (test data info must NOT influence the scaler)
# ══════════════════════════════════════════════════════════

scaler = StandardScaler()   # create scaler object (not yet fitted to any data)

X_train_scaled = scaler.fit_transform(X_train)  # learn mean/std from train data AND scale it
X_test_scaled  = scaler.transform(X_test)        # scale test data using SAME mean/std from train (no refit)

print("Step 6 — Feature scaling done (StandardScaler)")
# verify: after scaling, bmi column should have mean~0 and std~1
print(f"   Train bmi col — mean: {X_train_scaled[:,2].mean():.4f}, std: {X_train_scaled[:,2].std():.4f}")
print("=" * 55)
```

```python
# STEP 7 : Build Base AdaBoost Model (manual default params)
# AdaBoost trains many weak learners sequentially.
# Each new learner focuses more on samples the previous ones got wrong.
# Final prediction = weighted average of all weak learners.
# ══════════════════════════════════════════════════════════

# create a shallow decision tree — this is the weak learner inside AdaBoost
# max_depth=3 means tree can only make 3 levels of splits (intentionally simple/weak)
base_tree = DecisionTreeRegressor(max_depth=1, random_state=42)

ada_base = AdaBoostRegressor(
    estimator=base_tree,   # use the shallow tree as the weak learner
    n_estimators=100,      # train 100 weak learners one after another
    learning_rate=1.0,     # weight each tree contributes — 1.0 means full contribution
    loss="linear",         # how sample weights are updated after each tree: linear = proportional to error
    random_state=42        # fixed seed for reproducibility
)
```

```python
ada_base.fit(X_train_scaled, y_train)   # train AdaBoost on scaled training data

print("Step 7 — Base AdaBoost model trained")
print(f"   Estimators used : {len(ada_base.estimators_)}")   # actual number of trees trained (may be < 100 if early stop)
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 8 : Evaluate Base Model on Test Set
# MAE  = mean absolute error → avg |actual - predicted|
# RMSE = root mean squared error → penalises large errors more
# R2   = coefficient of determination
#        1.0 = perfect, 0 = as good as predicting mean, <0 = terrible
# ══════════════════════════════════════════════════════════

y_pred_base = ada_base.predict(X_test_scaled)   # generate predictions on test set (unseen data)

mae_base  = mean_absolute_error(y_test, y_pred_base)               # average absolute error
rmse_base = np.sqrt(mean_squared_error(y_test, y_pred_base))       # square root of mean squared error
r2_base   = r2_score(y_test, y_pred_base)                          # how much variance the model explains

print("Step 8 — Base Model Metrics")
print(f"   MAE  : {mae_base:.4f}")    # lower is better
print(f"   RMSE : {rmse_base:.4f}")   # lower is better
print(f"   R2   : {r2_base:.4f}")     # higher is better (max = 1.0)
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 9 : Hyperparameter Tuning with RandomizedSearchCV
# Hyperparameters = settings you choose before training (not learned)
# GridSearch tries EVERY combination → very slow
# RandomizedSearch tries N random combinations → much faster
# CV (cross-validation) = split train data into 5 folds, train on 4 test on 1, repeat 5 times
# ══════════════════════════════════════════════════════════

# define the search space — what values can each hyperparameter take?
param_dist = {
    "n_estimators"               : randint(50, 500),       # try any integer between 50 and 500
    "learning_rate"              : uniform(0.01, 1.5),     # try any float between 0.01 and 1.51
    "loss"                       : ["linear", "square", "exponential"],   # try all 3 loss options
    "estimator__max_depth"       : randint(1, 6),          # base tree depth: 1 to 5 (__ accesses base estimator param)
    "estimator__min_samples_split": randint(2, 20)         # min samples needed to split a node: 2 to 19
}
```

```python
base_tree_tune = DecisionTreeRegressor(random_state=42)    # fresh base tree for tuning (separate from base model)

ada_tune = AdaBoostRegressor(estimator=base_tree_tune, random_state=42)   # AdaBoost model to be tuned

random_search = RandomizedSearchCV(
    estimator=ada_tune,              # model to tune
    param_distributions=param_dist,  # search space defined above
    n_iter=30,                       # try 30 random combinations (30 × 5 folds = 150 total fits)
    scoring="r2",                    # use R2 as the metric to decide which params are best
    cv=5,                            # 5-fold cross-validation for each combination
    verbose=1,                       # print progress during search
    random_state=42,                 # fixed seed so same 30 combinations are picked every run
    n_jobs=-1                        # use all available CPU cores to speed up search
)
```

```python
print("Step 9 — RandomizedSearchCV starting (30 combos x 5 folds) ...")
random_search.fit(X_train_scaled, y_train)   # run the search on training data
```

```python
print()
print("   Best parameters found:")
for k, v in random_search.best_params_.items():    # loop through and print each best param
    print(f"      {k:45s}: {v}")
print(f"   Best CV R2 : {random_search.best_score_:.4f}")   # best average R2 across 5 folds
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 10 : Extract the Best Tuned Model
# best_estimator_ is already re-trained on FULL train data
# using the best hyperparameters found by RandomizedSearch
# ══════════════════════════════════════════════════════════

best_ada = random_search.best_estimator_   # extract the best model (already fitted on full train set)

print("Step 10 — Best model extracted from search")
print("=" * 55)
```

```python
# ══════════════════════════════════════════════════════════
# STEP 11 : Evaluate Tuned Model on Test Set
# Compare tuned model vs base model side by side
# ══════════════════════════════════════════════════════════

y_pred_tuned = best_ada.predict(X_test_scaled)   # predict using the tuned model on test data

mae_tuned  = mean_absolute_error(y_test, y_pred_tuned)           # MAE of tuned model
rmse_tuned = np.sqrt(mean_squared_error(y_test, y_pred_tuned))   # RMSE of tuned model
r2_tuned   = r2_score(y_test, y_pred_tuned)                      # R2 of tuned model

print("Step 11 — Tuned Model Metrics")
print(f"   MAE  : {mae_tuned:.4f}   (Base was {mae_base:.4f})")    # compare with base
print(f"   RMSE : {rmse_tuned:.4f}   (Base was {rmse_base:.4f})")  # compare with base
print(f"   R2   : {r2_tuned:.4f}   (Base was {r2_base:.4f})")      # compare with base
print("=" * 55)
```
