```
# pip install nltk spacy textblob contractions emoji unidecode beautifulsoup4
# python -m spacy download en_core_web_sm

import nltk
import re
import string
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```

# 1. TOKENIZATION
# Breaking a sentence into smaller pieces (words or sentences)


# --- Example A: Word tokenization ---
# Split a sentence into individual words

text = "I love learning NLP. It is very useful!"
tokens = word_tokenize(text)
print("Example A — Word Tokenization")
print(f"  Input : {text}")
print(f"  Output: {tokens}")
```

```
# --- Example B: Sentence tokenization ---
# Split a paragraph into individual sentences

text = "Python is great. I use it every day. It makes NLP easy!"
sentences = sent_tokenize(text)
print("\nExample B — Sentence Tokenization")
print(f"  Input : {text}")
print(f"  Output: {sentences}")
```

```
# --- Example C: Simple split tokenization ---
# The simplest way — just split by spaces (no library needed)

text = "Data science is fun"
tokens = text.split()
print("\nExample C — Simple Split")
print(f"  Input : {text}")
print(f"  Output: {tokens}")
print()
```

```
# 2. LOWERCASING
# Convert all text to small letters so "Apple" and "apple" are treated the same

# --- Example A: Lowercase a full sentence ---

text = "Machine Learning Is AWESOME!"
lowercase_text = text.lower()
print("Example A — Lowercase a sentence")
print(f"  Input : {text}")
print(f"  Output: {lowercase_text}")
```

```
# --- Example B: Lowercase a list of words ---
# Useful when you have already tokenized words

words = ['Hello', 'WORLD', 'Python', 'NLP', 'AI']
lowercase_words = [word.lower() for word in words]
print("\nExample B — Lowercase a list of words")
print(f"  Input : {words}")
print(f"  Output: {lowercase_words}")
```

```
# --- Example C: Lowercase before comparing two words ---
# "Hello" and "hello" are equal after lowercasing

word1 = "Hello"
word2 = "hello"
print("\nExample C — Compare words after lowercasing")
print(f"  Are '{word1}' and '{word2}' equal normally?      {word1 == word2}")
print(f"  Are they equal after lowercasing? {word1.lower() == word2.lower()}")

print()
```

```
# 3. STOPWORDS REMOVAL
# Remove common words like "is", "the", "a" that don't carry much meaning

stop_words = set(stopwords.words('english'))

# --- Example A: Remove stopwords from a sentence ---

text = "This is a sample sentence showing stopwords removal"
tokens = word_tokenize(text.lower())
filtered = [word for word in tokens if word not in stop_words]
print("Example A — Basic stopwords removal")
print(f"  Input : {text}")
print(f"  Output: {filtered}")
```

```
# --- Example B: Show which words were removed vs kept ---

text = "I am going to the market and buying some apples"
tokens = word_tokenize(text.lower())
removed = [word for word in tokens if word in stop_words]
kept    = [word for word in tokens if word not in stop_words]
print("\nExample B — See what was removed vs kept")
print(f"  Input   : {text}")
print(f"  Removed : {removed}")
print(f"  Kept    : {kept}")
```

```
# --- Example C: Add your own custom stopwords ---
# Sometimes you want to also remove extra domain-specific words

text = "please kindly note that the deadline is tomorrow"
tokens = word_tokenize(text.lower())
custom_stops = stop_words | {'please', 'kindly', 'note'}
filtered = [word for word in tokens if word not in custom_stops]
print("\nExample C — Custom stopwords")
print(f"  Input : {text}")
print(f"  Output: {filtered}")
```

```
# 4. PUNCTUATION REMOVAL
# Remove symbols like ! ? . , ; from text

# --- Example A: Remove punctuation using string.punctuation ---

text = "Hello! How are you? I'm doing great."
clean_text = text.translate(str.maketrans('', '', string.punctuation))
print("Example A — Using string.punctuation")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Remove punctuation using regex ---

text = "NLP is amazing!!! Don't you think so???"
clean_text = re.sub(r'[^\w\s]', '', text)
print("\nExample B — Using regex")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Keep only alphabetic tokens after tokenizing ---

text = "Wow! Python, NLP, and AI are great."
tokens = word_tokenize(text)
no_punct = [word for word in tokens if word.isalpha()]
print("\nExample C — Keep only alphabetic tokens")
print(f"  Input : {text}")
print(f"  Output: {no_punct}")
```

```
# 5. STEMMING
# Cut words down to their root form by removing endings (fast but rough)
# "running" -> "run",  "easily" -> "easili"  (result may not be a real word)

ps = PorterStemmer()

# --- Example A: Stem a list of related words ---

words = ['running', 'runs', 'easily', 'fairly', 'studies']
stemmed = [ps.stem(word) for word in words]
print("Example A — Stem a word list")
print(f"  Input : {words}")
print(f"  Output: {stemmed}")
```

```
# --- Example B: Stem every word in a sentence ---

sentence = "The cats are sitting and playing in the garden"
tokens = word_tokenize(sentence)
stemmed_sentence = ' '.join([ps.stem(word) for word in tokens])
print("\nExample B — Stem a full sentence")
print(f"  Input : {sentence}")
print(f"  Output: {stemmed_sentence}")
```

```
# --- Example C: Stem different tenses of the same verb ---

words = ['eat', 'eating', 'eats', 'eaten']
stemmed = [ps.stem(word) for word in words]
print("\nExample C — Different tenses of one verb")
print(f"  Input : {words}")
print(f"  Output: {stemmed}")
```

```
# 6. LEMMATIZATION
# Convert words to their proper dictionary form (slower but more accurate)
# "running" -> "run",  "ran" -> "run",  "cats" -> "cat"

lemmatizer = WordNetLemmatizer()

# --- Example A: Lemmatize verbs ---
# pos='v' tells the lemmatizer to treat words as verbs

words = ['running', 'ran', 'studies', 'went', 'eating']
lemmatized = [lemmatizer.lemmatize(word, pos='v') for word in words]
print("Example A — Lemmatize verbs")
print(f"  Input : {words}")
print(f"  Output: {lemmatized}")
```

```
# --- Example B: Lemmatize nouns (plural -> singular) ---

words = ['cats', 'geese', 'children', 'teeth', 'buses']
lemmatized = [lemmatizer.lemmatize(word, pos='n') for word in words]
print("\nExample B — Lemmatize nouns")
print(f"  Input : {words}")
print(f"  Output: {lemmatized}")
```

```
# --- Example C: Compare stemming vs lemmatization side by side ---
# Lemmatization gives real words; stemming may give partial/non-real words

words = ['studies', 'fairly', 'crying']
print("\nExample C — Stemming vs Lemmatization comparison")
for word in words:
    stem  = ps.stem(word)
    lemma = lemmatizer.lemmatize(word, pos='v')
    print(f"  '{word}' -> Stem: '{stem}'  |  Lemma: '{lemma}'")
```

```
# 7. REMOVING NUMBERS
# Delete digits from text when numbers are not useful

# --- Example A: Remove all digits with regex ---

text = "I bought 5 apples and 10 oranges in 2024"
clean_text = re.sub(r'\d+', '', text).strip()
print("Example A — Remove digits using regex")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Remove digits character by character ---

text = "Python3 is better than Python2 for NLP"
clean_text = ''.join([ch for ch in text if not ch.isdigit()])
print("\nExample B — Remove digits character by character")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Keep only non-numeric tokens after tokenizing ---

text = "In 2024 the AI model scored 95 percent accuracy"
tokens = word_tokenize(text)
no_numbers = [word for word in tokens if not word.isdigit()]
print("\nExample C — Remove number tokens after tokenizing")
print(f"  Input : {text}")
print(f"  Output: {no_numbers}")
```

```
# 8. REMOVING URLs
# Delete web links like https://example.com from text

# --- Example A: Remove http and www links ---

text = "Visit https://www.google.com for search and check http://example.org too"
clean_text = re.sub(r'http\S+|www\.\S+', '', text).strip()
print("Example A — Remove http and www links")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Remove URL then clean the extra blank spaces left behind ---

text = "Read more at https://news.com/article and https://blog.org/post today"
clean_text = re.sub(r'http\S+|www\.\S+', '', text)
clean_text = ' '.join(clean_text.split())
print("\nExample B — Remove URL then clean spaces")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Replace URL with a placeholder word ---
# Useful when you want the model to know a link existed

text = "Check this link https://example.com for details"
clean_text = re.sub(r'http\S+|www\.\S+', 'URL', text)
print("\nExample C — Replace URL with placeholder")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# 9. REMOVING SPECIAL CHARACTERS
# Remove @, #, $, %, ^, & etc. — keep only letters, numbers, and spaces

# --- Example A: Remove all non-alphanumeric characters ---

text = "Hello @world! #NLP is $amazing & @awesome!!!"
clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', text).strip()
print("Example A — Remove all special characters")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Remove hashtags and @mentions (social media cleaning) ---

text = "Follow @john and read #MachineLearning posts today!"
clean_text = re.sub(r'[@#]\w+', '', text)
clean_text = ' '.join(clean_text.split())
print("\nExample B — Remove hashtags and @mentions")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Keep letters and spaces only (strictest cleaning) ---

text = "AI & ML are #1 in tech! 100% sure."
clean_text = re.sub(r'[^a-zA-Z\s]', '', text)
clean_text = ' '.join(clean_text.split())
print("\nExample C — Keep letters and spaces only")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# 10. REMOVING EXTRA WHITESPACE
# Remove double spaces and trim spaces from the start/end

# --- Example A: Remove all extra spaces at once ---

text = "  Hello    world   how   are   you  "
clean_text = ' '.join(text.split())
print("Example A — Remove all extra spaces")
print(f"  Input : '{text}'")
print(f"  Output: '{clean_text}'")
```

```
# --- Example B: Remove only leading and trailing spaces ---
# .strip() only removes spaces at the edges, not in the middle

text = "   Hello world   "
clean_text = text.strip()
print("\nExample B — Strip edges only (.strip)")
print(f"  Input : '{text}'")
print(f"  Output: '{clean_text}'")
```

```
# --- Example C: Replace multiple spaces with one space using regex ---

text = "NLP  is   very    powerful"
clean_text = re.sub(r'\s+', ' ', text).strip()
print("\nExample C — Replace multiple spaces with one")
print(f"  Input : '{text}'")
print(f"  Output: '{clean_text}'")
```

```
# 11. REMOVING HTML TAGS
# Strip out HTML code like <p>, <b>, <div> from text (useful for web data)

from bs4 import BeautifulSoup   # pip install beautifulsoup4

# --- Example A: Strip basic HTML tags using BeautifulSoup ---

html_text = "<p>This is <b>bold</b> and <i>italic</i> text!</p>"
clean_text = BeautifulSoup(html_text, 'html.parser').get_text()
print("Example A — Remove basic HTML tags")
print(f"  Input : {html_text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Remove HTML using regex (no library needed) ---

html_text = "<h1>Welcome</h1><p>This is a <a href='#'>link</a>.</p>"
clean_text = re.sub(r'<[^>]+>', '', html_text)
print("\nExample B — Remove HTML tags using regex")
print(f"  Input : {html_text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Extract text from nested HTML with a space separator ---

html_text = "<div><span>Hello</span><span>World</span><br/>How are you?</div>"
clean_text = BeautifulSoup(html_text, 'html.parser').get_text(separator=' ')
print("\nExample C — Extract text from nested HTML")
print(f"  Input : {html_text}")
print(f"  Output: {clean_text}")
```

```
# 12. EXPANDING CONTRACTIONS
# Convert short forms to full forms: "don't" -> "do not", "I'm" -> "I am"

import subprocess
subprocess.run(['pip', 'install', 'contractions', '-q'])
import contractions

# --- Example A: Expand contractions in a sentence ---

text = "I don't think it's a good idea. We can't do this."
expanded = contractions.fix(text)
print("Example A — Expand contractions in a sentence")
print(f"  Input : {text}")
print(f"  Output: {expanded}")
```

```
# --- Example B: Expand I'm / you're / it's ---

text = "I'm happy that you're learning NLP. It's really useful."
expanded = contractions.fix(text)
print("\nExample B — Expand I'm / you're / it's")
print(f"  Input : {text}")
print(f"  Output: {expanded}")
```

```
# --- Example C: Expand first, then tokenize for cleaner results ---
# "don't" might split wrongly without expanding first

text = "She wouldn't have known if they'd told her."
expanded = contractions.fix(text)
tokens = word_tokenize(expanded)
print("\nExample C — Expand then tokenize")
print(f"  Input    : {text}")
print(f"  Expanded : {expanded}")
print(f"  Tokens   : {tokens}")
```

```
# 13. REMOVING EMOJIS
# Strip out emoji symbols from text

import subprocess
subprocess.run(['pip', 'install', 'emoji', '-q'])
import emoji

# --- Example A: Remove emojis from a product review ---

text = "I love this product! 😊 It's amazing! 🚀"
clean_text = emoji.replace_emoji(text, replace='').strip()
print("Example A — Remove emojis from a review")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Replace emojis with a placeholder word ---
# Keeps the information that an emoji was present

text = "Great food 😋 but slow delivery 😞"
clean_text = emoji.replace_emoji(text, replace='EMOJI')
print("\nExample B — Replace emoji with EMOJI placeholder")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Remove emojis using regex (no library needed) ---
# Removes all non-ASCII characters which covers most emojis

text = "Hello 😀 World 🌍 Let us go 🎉"
clean_text = re.sub(r'[^\x00-\x7F]+', '', text).strip()
print("\nExample C — Remove emojis with regex")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# 14. SPELL CORRECTION
# Automatically fix typos and misspelled words

from textblob import TextBlob   # pip install textblob

# --- Example A: Fix a sentence with several typos ---

text = "I havve a quikc questoin abuot machin lerning"
corrected = str(TextBlob(text).correct())
print("Example A — Fix multiple typos")
print(f"  Input : {text}")
print(f"  Output: {corrected}")
```

```
# --- Example B: Fix casual typing mistakes ---

text = "ths is an amazng sentance wth erors"
corrected = str(TextBlob(text).correct())
print("\nExample B — Fix casual typing mistakes")
print(f"  Input : {text}")
print(f"  Output: {corrected}")
```

```
# --- Example C: See which words were corrected one by one ---

text = "Naturel languege procesing is intresting"
blob = TextBlob(text)
print("\nExample C — Word by word correction")
print(f"  Input : {text}")
for word in blob.words:
    fixed = str(TextBlob(word).correct())
    if fixed != word:
        print(f"  '{word}' -> '{fixed}'")
print(f"  Full output: {str(blob.correct())}")
```

```
# 15. REMOVING ACCENTS (Normalizing Characters)
# Convert accented letters like é, ñ, ü to their plain English equivalents


import subprocess
subprocess.run(['pip', 'install', 'unidecode', '-q'])
from unidecode import unidecode

# --- Example A: Remove accents from French-style words ---

text = "Héllo wörld! Café naïve résumé"
clean_text = unidecode(text)
print("Example A — French accented characters")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example B: Normalize Spanish characters ---

text = "El niño jugó en el jardín con múcha alegría"
clean_text = unidecode(text)
print("\nExample B — Spanish accented characters")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
# --- Example C: Remove accents using built-in unicodedata (no extra library) ---
# Works without installing unidecode

import unicodedata
text = "Héllo Wörld Über café"

# Step 1: Decompose letters — e.g. é becomes e + accent mark
# Step 2: Drop the accent mark characters (unicode category 'Mn')

clean_text = unicodedata.normalize('NFD', text)
clean_text = ''.join(ch for ch in clean_text if unicodedata.category(ch) != 'Mn')
print("\nExample C — Using built-in unicodedata (no install needed)")
print(f"  Input : {text}")
print(f"  Output: {clean_text}")
```

```
print("1.  Tokenization          — Always first step in NLP")
print("2.  Lowercasing           — When case doesn't matter")
print("3.  Stopwords Removal     — Classification, topic modeling")
print("4.  Punctuation Removal   — Word frequency analysis")
print("5.  Stemming              — When speed matters over accuracy")
print("6.  Lemmatization         — When accuracy matters (chatbots, QA)")
print("7.  Remove Numbers        — When digits are noise")
print("8.  Remove URLs           — Social media / web scraped text")
print("9.  Remove Special Chars  — Cleaning messy raw text")
print("10. Remove Whitespace     — Always good to do")
print("11. Remove HTML Tags      — Web scraped data")
print("12. Expand Contractions   — Before tokenization for better results")
print("13. Remove Emojis         — When emojis are noise (or keep for sentiment)")
print("14. Spell Correction      — User-generated content, reviews")
print("15. Remove Accents        — Multilingual text normalization")





1.  Remove HTML Tags          → Clean raw web/scraped data first
2.  Remove URLs               → Remove links before any splitting
3.  Expand Contractions       → "don't" → "do not" (before tokenizing)
4.  Remove Emails & Mentions  → Clean @user, email@gmail.com
5.  Remove Special Characters → Remove @, #, $, % etc.
6.  Remove Numbers            → Remove digits if not needed
7.  Lowercasing               → Convert everything to small letters
8.  Tokenization              → NOW split into words
9.  Remove Stopwords          → Remove "is", "the", "a" etc.
10. Remove Punctuation        → Remove leftover . , ! ? tokens
11. Spell Correction          → Fix typos (optional, slow)
12. Stemming OR Lemmatization → Reduce words to root form
13. Remove Extra Whitespace   → Final cleanup
14. Remove Accents            → Normalize é → e, ñ → n
15. Remove Emojis             → If emojis are noise
```
