# **Faster R-CNN**
Faster Region-based Convolutional Neural Network

```python
!pip install torchvision
import torchvision
import torch
from PIL import Image
import torchvision.transforms as T

# Pretrained model load
model = torchvision.models.detection.fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Image prepare
image = Image.open("/content/many at one.jpg")
transform = T.ToTensor()
img_tensor = transform(image)

# Predict
with torch.no_grad():
    predictions = model([img_tensor])

print(predictions[0]['labels'])   # detected class IDs
print(predictions[0]['scores'])   # confidence scores
print(predictions[0]['boxes'])    # bounding boxes
```

# **Detectron2**

```python
!pip install pyyaml==5.1
!pip install torch torchvision
!pip install 'git+https://github.com/facebookresearch/detectron2.git'
```

```python
import torch
import cv2
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

from detectron2 import model_zoo
from detectron2.engine import DefaultPredictor
from detectron2.config import get_cfg
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog

# Config setup
cfg = get_cfg()
cfg.merge_from_file(model_zoo.get_config_file(
    "COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml"
))
cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5  # confidence threshold
cfg.MODEL.WEIGHTS = model_zoo.get_checkpoint_url(
    "COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml"
)
cfg.MODEL.DEVICE = "cuda"  # GPU use

# Predictor
predictor = DefaultPredictor(cfg)

# Image load
image = cv2.imread("/content/many at one.jpg")

# Predict
outputs = predictor(image)

# Result visualize
v = Visualizer(image[:, :, ::-1], MetadataCatalog.get(cfg.DATASETS.TRAIN[0]), scale=1.2)
out = v.draw_instance_predictions(outputs["instances"].to("cpu"))

# Display karo
plt.figure(figsize=(12, 8))
plt.imshow(out.get_image())
plt.axis("off")
plt.show()
```

# **SSD MobileNet (TensorFlow)**

```python
!pip install tensorflow
!pip install tensorflow-hub
```

```python
import tensorflow as tf
import tensorflow_hub as hub
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Pretrained SSD MobileNet V2 load (TF Hub se)
model = hub.load("https://tfhub.dev/tensorflow/ssd_mobilenet_v2/2")

# Image load and prepare
def load_image(path):
    image = Image.open(path).convert("RGB")
    image = np.array(image)
    return image

image = load_image("/content/many at one.jpg")
input_tensor = tf.convert_to_tensor(image)[tf.newaxis, ...]

# Predict
results = model(input_tensor)

# Results extract
boxes   = results['detection_boxes'][0].numpy()
scores  = results['detection_scores'][0].numpy()
classes = results['detection_classes'][0].numpy().astype(int)

# show only high confidence detections
threshold = 0.5
for i in range(len(scores)):
    if scores[i] >= threshold:
        print(f"Class: {classes[i]}, Confidence: {scores[i]:.2f}, Box: {boxes[i]}")

# Bounding boxes draw
fig, ax = plt.subplots(1, figsize=(12, 8))
ax.imshow(image)

h, w, _ = image.shape
for i in range(len(scores)):
    if scores[i] >= threshold:
        ymin, xmin, ymax, xmax = boxes[i]
        x, y = xmin * w, ymin * h
        bw, bh = (xmax - xmin) * w, (ymax - ymin) * h
        rect = plt.Rectangle((x, y), bw, bh,
                              linewidth=2, edgecolor='red', facecolor='none')
        ax.add_patch(rect)
        ax.text(x, y, f"{classes[i]}: {scores[i]:.2f}",
                color='white', fontsize=10,
                bbox=dict(facecolor='red', alpha=0.5))

plt.axis("off")
plt.show()
```
