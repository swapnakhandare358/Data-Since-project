# **BOW**

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "I love python language",
    "I love machine learning",
    "python is easy to learn",
]
```

```python
# Initialize CountVectorizer (converts text to word count matrix)
cv = CountVectorizer()
cv.fit(texts)
```

```python
print(cv.get_feature_names_out())
```

```python
result = cv.transform(texts)
```

```python
print(result.toarray())
```

```python
new = ["I love learning"]
```

```python
vector = cv.transform(new)
print(vector.toarray())
```

# **TF-IDF**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

texts = [
    "I love python",
    "I love machine learning",
    "python is easy to learn",
]
```

```python
tfidf = TfidfVectorizer()
tfidf.fit(texts)
```

```python
print(tfidf.get_feature_names_out())
```

```python
result = tfidf.transform(texts)
```

```python
print(result.toarray().round(2))
```

```python
bow = cv.fit_transform(texts).toarray()  # Get Bag of Words matrix
tfidf_vec = tfidf.transform(texts).toarray().round(2)
```

```python
print("BoW:  ", bow[0])      # Simple counts (0 or 1 here)
print("TFIDF:", tfidf_vec[0])
```

```python
new = ["python is great"]

# Transform new text using fitted TF-IDF vectorizer
vector = tfidf.transform(new)

print(vector.toarray().round(2))
```

# **N-grams**

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "I love machine learning",
    "machine learning is fun",
]
```

```python
# Unigram: single words only
cv1 = CountVectorizer(ngram_range=(1,1))

cv1.fit(texts)

print("Unigram:", cv1.get_feature_names_out())
```

```python
# Bigram: pairs of consecutive words (2 words together)
cv2 = CountVectorizer(ngram_range=(2,2))
cv2.fit(texts)
print("Bigram: ", cv2.get_feature_names_out())
```

```python
# Trigram: triplets of consecutive words (3 words together)
cv3 = CountVectorizer(ngram_range=(3,3))
cv3.fit(texts)
print("Trigram:", cv3.get_feature_names_out())
```

```python
from gensim.models import Word2Vec

# Tokenized sentences (each sentence is a list of words)
sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "easy"],
    ["deep", "learning", "is", "fun"],
    ["cat", "and", "dog", "are", "animals"],
    ["cat", "drinks", "milk"],
    ["dog", "eats", "food"],
]
```

```python
model = Word2Vec(
    sentences,
    vector_size=10,   # Dimensionality of word vectors (10 numbers per word)
    window=4,         # Context window size (look at 2 words before/after)
    min_count=1,      # Include words that appear at least once
    sg=0              # 0=CBOW (predict word from context), 1=SkipGram (predict context from word)
)
```

```python
vector = model.wv["python"]
```

```python
print(vector)  # 10 numbers representing meaning of "python"
print(vector.shape)
```

```python
vector = model.wv["machine"]
print(vector)
print(vector.shape)
```

```python
similar = model.wv.most_similar("cat", topn=3)
print(similar)
```

```python
score = model.wv.similarity("cat", "dog")
print(score)
```

```python
score = model.wv.similarity("cat", "animals")
print(score)
```

# **FastText**

```python
from gensim.models import FastText

corpus = [
    ["i", "love", "python"],
    ["python", "is", "easy"],
    ["cat", "and", "dog", "are", "animals"],
    ["cat", "drinks", "milk"],
]
```

```python
model = FastText(
    corpus,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0  # 1=SkipGram (FastText works better with SkipGram)
)
```

```python
print(model.wv["cat"])                          # Vector for known word
print(model.wv.most_similar("cat", topn=3))    # Similar words
print(model.wv.similarity("cat", "dog"))       # Similarity score
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

reviews = [
    "battery life is excellent",
    "phone battery drains very fast",
    "camera quality is amazing",
    "screen display is very bright",
    "worst phone i ever bought",
]
```

```python
vectors = model.encode(reviews)
```

```python
query = "battery is good"

query_vec = model.encode([query])
```

```python
scores = cosine_similarity(query_vec, vectors)[0]
```

```python
# Display similarity scores (higher = more similar)
for i, score in enumerate(scores):
    print(f"{score:.2f} — {reviews[i]}")
```
